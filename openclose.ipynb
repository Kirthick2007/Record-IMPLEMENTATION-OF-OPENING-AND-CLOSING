```python
# Import the necessary packages
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
def load_img():
    blank_img =np.zeros((200,200))
    font = cv2.FONT_HERSHEY_SIMPLEX
    cv2.putText(blank_img,text='KI',org=(10,175), fontFace=font,fontScale= 5,color=(255,255,255),thickness=25,lineType=cv2.LINE_AA)
    return blank_img
```

```python
def display_img(img):
    fig = plt.figure(figsize=(12,10))
    ax = fig.add_subplot(111)
    ax.imshow(img,cmap='gray')
    plt.show()
```

```python
img = load_img()
```

```python
display_img(img)
```

```python
img=load_img()
```

```python
black_noise = np.random.randint(low=0,high=2,size=(200,200))
```

```python
black_noise
```

```python
black_noise= black_noise * -255
```

```python
black_noise_img = img + black_noise
```

```python
black_noise_img     # display
```

```python
black_noise_img[black_noise_img==-255] = 0
```

```python
display_img(black_noise_img)
```

```python
kernel = np.ones((7,7), dtype=np.uint8)
```

```python
closing = cv2.morphologyEx(black_noise_img, cv2.MORPH_CLOSE, kernel)
```

```python
display_img(closing)
```

```python
white_noise = np.random.randint(low=0,high=2,size=(200,200))
white_noise = white_noise * 255
noise_img = white_noise + img
display_img(noise_img)
```

```python
kernel = np.ones((5,5), dtype=np.uint8)
opening = cv2.morphologyEx(noise_img,cv2.MORPH_OPEN,kernel)
display_img(opening)
```
